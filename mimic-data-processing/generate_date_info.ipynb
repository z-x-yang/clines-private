```python
import pandas as pd
path = "/home/zoy043/Datasets/NOTEEVENTS.csv.gz"
notes = pd.read_csv(path, compression='gzip')
```

```python
notes = notes[notes['CATEGORY']=='Discharge summary'][['ROW_ID', 'TEXT']]
```

```python
def parse_mask(strings):
    import re
    def subpattern(matchobj):
        return ' '+matchobj.group(0).lstrip('[*').rstrip('*]')+' '
    strings = re.sub(r"\[+\*+[0-9- ]*\*+\]+", subpattern, strings)
    strings = re.sub(r"\[+\*+Date[0-9- ]*\*+\]+", subpattern, strings)
    strings = re.sub(r"\[\*\*[/0-9A-Za-z ]*Month[/0-9A-Za-z ]*\*\*\]", subpattern, strings)
    strings = re.sub(r"\[+\*+Hospital[0-9- ]*\*+\]+", "____", strings)
    strings = re.sub(r"\[+\*+[0-9A-Za-z()\- ]*[Nn]ame[0-9A-Za-z()\- ]*\*+\]+", "____", strings)
    strings = re.sub(r"\[+\*+[0-9A-Za-z()\- ]*Location[0-9A-Za-z()\- ]*\*+\]+", "____", strings)
    strings = re.sub(r"\[+\*+[0-9A-Za-z()\- ]*Company[0-9A-Za-z()\- ]*\*+\]+", "____", strings)
    strings = re.sub(r"\[+\*+[0-9A-Za-z()\- ]*Country[0-9A-Za-z()\- ]*\*+\]+", "____", strings)
    strings = re.sub(r"\[+\*+[0-9A-Za-z()\- ]*Address[0-9A-Za-z()\- ]*\*+\]+", "____", strings)
    strings = re.sub(r"\[+\*+[0-9A-Za-z()\- ]*Tele[\[\]0-9A-Za-z()\- ]*\*+\]+", "____", strings)
    strings = re.sub(r"\[+\*+[0-9A-Za-z()\- ]*Telephone.*\*+\]+", "____", strings)
    strings = re.sub(r"\[+\*+[0-9A-Za-z()\- ]*[0-9A-Za-z()\- ]*\*+\]+", "____", strings)
    return strings
# parse_mask('[***2378-29-392***],hdkasbhala[***2839-29-392***]hsdkjahkjsdalk[***8923-29-392***]')
```

```python
notes['TEXT'] = notes['TEXT'].apply(parse_mask)
```
