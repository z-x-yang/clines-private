```python
from tqdm import tqdm

path = "/home/zoy043/Datasets/MRCONSO.RRF"

with open(path, "r") as f:
    lines = f.readlines()

cui_term = set()
for l in tqdm(lines):
    lst = l.rstrip("\n").split("|")
    cui, lang, synonym, sty = lst[0], lst[1], lst[14], lst[11]
    if lang != 'ENG': continue
    cui_term.update([cui+'||'+synonym.lower()])
cui_term = list(cui_term)
print(len(cui_term))

with open('./umls_dictionary.txt', 'w') as f:
    for item in cui_term:
        f.write(item+'\n')
```
