```python
from tqdm import tqdm
import pandas as pd

semantic_type = set(['T005','T007','T017','T022','T031','T033','T037','T038','T058','T062','T074',
                    'T082','T091','T092','T097','T098','T103','T168','T170','T201','T204'])
semantic_type_ontology = pd.read_csv('./STY.csv')
semantic_type_size = 0
while len(semantic_type)!=semantic_type_size:
    semantic_type_size = len(semantic_type)
    for i in range(len(semantic_type_ontology)):
        if semantic_type_ontology['Parents'][i][-4:] in semantic_type:
            semantic_type.update([semantic_type_ontology['Class ID'][i][-4:]])
source_onto = ['CPT','FMA','GO','HGNC','HPO','ICD10','ICD10CM','ICD9CM','MDR','MSH','MTH',
                'NCBI','NCI','NDDF','NDFRT','OMIM','RXNORM','SNOMEDCT_US']

cuiset = dict()
path = "/n/data1/hsph/biostat/celehs/lab/SHARE/From_Keming/UMLS/umls2020/MRSTY.RRF"
with open(path, "r") as f:
    mrsty = f.readlines()
for l in mrsty:
    lst = l.rstrip("\n").split("|")
    cui, sty, string = lst[0], lst[1], lst[3]
    if sty in semantic_type:
        cuiset[cui] = lst[3]
        

path = "/n/data1/hsph/biostat/celehs/lab/SHARE/From_Keming/UMLS/umls2020/MRCONSO.RRF"
with open(path, "r") as f:
    lines = f.readlines()


cui_term = set()
for l in tqdm(lines):
    lst = l.rstrip("\n").split("|")
    cui, lang, synonym, onto = lst[0], lst[1], lst[14], lst[11]
    if lang != 'ENG' or onto not in source_onto or cui not in cuiset: continue
    cui_term.update([cui+'||'+synonym.lower()+'||'+cuiset[cui]])
cui_term = list(cui_term)
print(len(cui_term))

with open('./umls_dictionary.txt', 'w') as f:
    for item in cui_term:
        f.write(item+'\n')
```
